# Классификатор тональности

Цель: научить модель различать позитивные и негативные сообщения.

## 1. Подготовка библиотек

```python
# ШАГ 01 — Пути к файлам

# Path — инструмент Python для работы с путями к папкам и файлам.
from pathlib import Path

# Pandas понадобится для чтения CSV и работы с таблицами.
# Имя pd — принятое короткое обозначение pandas.
import pandas as pd

# Указываем папку, в которой лежат исходные данные.
DATA_DIR = Path("data")

# Составляем путь к каждому файлу.
# Знак / здесь соединяет путь к папке с именем файла.
NEGATIVE_FILE = DATA_DIR / "negative.csv"
POSITIVE_FILE = DATA_DIR / "positive.csv"

# Показываем папку, из которой сейчас работает Jupyter.
print("Рабочая папка:", Path.cwd())

# exists() проверяет, найден ли файл по указанному пути.
print("Найден negative.csv:", NEGATIVE_FILE.exists())
print("Найден positive.csv:", POSITIVE_FILE.exists())
```

```python
# ШАГ 02 — Загрузка CSV и перевод меток

# Читаем из каждого CSV два нужных столбца:
# ttext — текст сообщения, ttype — исходная метка (-1 или 1).
negative = pd.read_csv(
    NEGATIVE_FILE,
    usecols=["ttext", "ttype"],
)

positive = pd.read_csv(
    POSITIVE_FILE,
    usecols=["ttext", "ttype"],
)

# Задаём правило перевода меток:
# -1 — негатив — превращаем в 0;
#  1 — позитив — оставляем как 1.
label_mapping = {-1: 0, 1: 1}

# Создаём в каждой таблице новый столбец label.
# Исходный столбец ttype при этом сохраняется.
negative["label"] = negative["ttype"].map(label_mapping)
positive["label"] = positive["ttype"].map(label_mapping)

# Проверяем количество строк с каждой новой меткой.
print("Метки в negative.csv:")
print(negative["label"].value_counts())

print("\nМетки в positive.csv:")
print(positive["label"].value_counts())

# Показываем первые три строки каждой таблицы.
# Так можно сравнить исходную метку ttype с новой label.
display(negative[["ttext", "ttype", "label"]].head(3))
display(positive[["ttext", "ttype", "label"]].head(3))
```

```python
# ШАГ 03 — Объединение данных

# Ставим строки negative и positive друг под другом.
# В каждой строке сохраняются текст, исходная метка и новая метка label.
df = pd.concat([negative, positive], ignore_index=True)

print("Строк после объединения:", len(df))
print("Сообщений каждого класса:")
print(df["label"].value_counts().sort_index())
```

```python
# ШАГ 04 — Функция проверки и удаления повторов

def remove_text_duplicates(data, text_column):
    print(f"\n=== Проверка столбца {text_column} ===")

    # Создаём копию таблицы без отсутствующих значений
    # в проверяемом столбце.
    working = data.dropna(subset=[text_column]).copy()
    missing_count = len(data) - len(working)

    # Находим строки, в которых текст пустой или состоит из пробелов.
    blank = working[text_column].astype(str).str.strip().eq("")
    blank_count = int(blank.sum())

    # Удаляем такие строки.
    working = working.loc[~blank].copy()
    print("1. Строк без текста:", missing_count + blank_count)

    # Выбираем столбцы, которые покажем в таблицах с примерами.
    columns_to_show = [text_column, "label"]

    # При проверке очищенного текста показываем также исходный текст.
    if text_column != "ttext" and "ttext" in working.columns:
        columns_to_show = ["ttext", text_column, "label"]

    # Проверяем, есть ли одинаковые тексты с разными метками:
    # один и тот же текст размечен и как негатив, и как позитив.
    conflicting = (
        working.groupby(text_column)["label"].transform("nunique") > 1
    )

    # Получаем список разных текстов с противоречивыми метками.
    conflicting_texts = (
        working.loc[conflicting, text_column].drop_duplicates()
    )

    print(
        "2. Разных текстов с противоречивыми метками:",
        len(conflicting_texts),
    )

    if len(conflicting_texts) > 0:
        print("Примеры конфликтов — все эти строки будут удалены:")

        # Берём первые три конфликтующих текста.
        examples = conflicting_texts.head(3)

        # Показываем все строки с этими текстами и их метки.
        display(
            working.loc[
                working[text_column].isin(examples),
                columns_to_show,
            ].sort_values([text_column, "label"])
        )

    # Удаляем все строки с противоречивыми метками.
    working = working.loc[~conflicting].copy()

    # Для каждой строки считаем, сколько всего строк
    # имеют такой же текст в проверяемом столбце.
    repeat_count = (
        working.groupby(text_column)[text_column].transform("size")
    )

    # Первую строку с данным текстом оставляем.
    # Все последующие одинаковые строки помечаем как повторы.
    duplicated = working.duplicated(
        subset=[text_column],
        keep="first",
    )
    duplicate_count = int(duplicated.sum())

    print("3. Всего повторных строк для удаления:", duplicate_count)

    if duplicate_count > 0:
        # Берём строки, которые будут удалены как повторы.
        duplicate_examples = (
            working.loc[duplicated, columns_to_show].copy()
        )

        # Если текст встретился 4 раза, одну строку оставляем,
        # а 3 копии удаляем. Это число добавляем в таблицу.
        duplicate_examples["удаляется_копий"] = (
            repeat_count.loc[duplicated].to_numpy() - 1
        )

        # Каждый повторяющийся текст показываем только один раз.
        # Выводим пять текстов с наибольшим числом удаляемых копий.
        duplicate_examples = (
            duplicate_examples
            .drop_duplicates(subset=[text_column])
            .sort_values("удаляется_копий", ascending=False)
            .head(5)
        )

        print(
            "Примеры повторяющихся текстов "
            "и число удаляемых копий:"
        )
        display(duplicate_examples)

    # Удаляем все строки, помеченные как повторы.
    working = working.loc[~duplicated].copy()

    print("Итог: строк осталось:", len(working))

    # Возвращаем обработанную таблицу.
    return working
```

```python
# ШАГ 05 — Применение функции к исходному тексту

# Передаём функции общую таблицу df и имя столбца ttext.
# Функция покажет результаты проверки и вернёт таблицу без повторов.
df = remove_text_duplicates(df, "ttext")
```

```python
# ШАГ 06 — Функция очистки текста

# re — стандартная библиотека Python для поиска и замены по шаблонам.
import re

def clean_text(text):
    # Превращаем входное значение в строку.
    # lower() переводит заглавные буквы в строчные.
    text = str(text).lower()

    # Удаляем ссылки, начинающиеся с http://, https:// или www.
    # Вместо ссылки ставим пробел, чтобы соседние слова не склеились.
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    # Удаляем упоминания пользователей вместе с символом @.
    text = re.sub(r"@\w+", " ", text)

    # Удаляем хештеги целиком: и символ #, и слово после него.
    text = re.sub(r"#\w+", " ", text)

    # Оставляем русские буквы, цифры и пробелы; остальные символы заменяем пробелом.
    text = re.sub(r"[^а-яё0-9\s]", " ", text)

    # Заменяем несколько пробелов подряд одним пробелом.
    # strip() убирает пробелы в начале и конце строки.
    text = re.sub(r"\s+", " ", text).strip()

    # Отдаём готовый текст туда, откуда вызвали функцию.
    return text
```

```python
# ШАГ 07 — Применение очистки ко всем сообщениям

# Берём столбец ttext с исходными сообщениями.
# map(clean_text) вызывает нашу функцию clean_text для каждого сообщения.
# Результаты записываем в новый столбец text_clean.
# Исходный столбец ttext сохраняется: он пригодится для сравнения.
df["text_clean"] = df["ttext"].map(clean_text)

# Проверяем, сколько сообщений находится в таблице.
print("Всего сообщений:", len(df))

# После удаления ссылок, упоминаний и латиницы некоторые тексты
# могли стать пустыми. Считаем такие строки, но пока не удаляем.
print("Пустых текстов после очистки:", df["text_clean"].eq("").sum())

# Показываем пять примеров: исходный текст, очищенный текст и метку.
# random_state=42 позволяет при повторном запуске видеть те же примеры.
display(
    df[["ttext", "text_clean", "label"]].sample(5, random_state=42)
)

# Ищем сообщения, в которых есть хотя бы одно слово или число
# длиной от двух символов — такие признаки сможет взять TF-IDF.
has_usable_token = df["text_clean"].str.contains(
    r"\b[а-яё0-9]{2,}\b",
    regex=True,
)

print("Текстов без пригодных для TF-IDF слов:", (~has_usable_token).sum())

# Показываем примеры, включая исходный текст.
display(
    df.loc[~has_usable_token, ["ttext", "text_clean", "label"]].head(10)
)
```

```python
# ШАГ 08 — Удаление непригодных очищенных текстов

# Запоминаем число строк до удаления, чтобы оценить результат.
rows_before = len(df)

# Стандартный TF-IDF берёт слова длиной от двух символов.
# Если в text_clean нет ни одного такого слова или числа,
# модель не получит из этого сообщения полезных признаков.
# na=False считает отсутствующее значение непригодным.
has_usable_token = df["text_clean"].str.contains(
    r"\b[а-яё0-9]{2,}\b",
    regex=True,
    na=False,
)

# Оставляем только строки, в которых найден подходящий токен.
# copy() создаёт отдельную таблицу с оставшимися строками.
df = df.loc[has_usable_token].copy()

print("Строк до проверки:", rows_before)
print("Удалено строк без пригодных токенов:", rows_before - len(df))
print("Строк после проверки:", len(df))

# Теперь проверяем повторы уже по очищенному тексту.
# Функция также уберёт тексты с противоречивыми метками.
df = remove_text_duplicates(df, "text_clean")
```
