Классификатор тональности

Цель: научить модель различать позитивные и негативные сообщения.

## 1. Подготовка библиотек

```python
# ШАГ 01 — Пути к файлам

# Path — инструмент Python для работы с путями к папкам и файлам.
from pathlib import Path

# Pandas понадобится для чтения CSV и работы с таблицами.
# Имя pd — принятое короткое обозначение pandas.
import pandas as pd

# Указываем папку, в которой лежат исходные данные.
DATA_DIR = Path("data")

# Составляем путь к каждому файлу.
# Знак / здесь соединяет путь к папке с именем файла.
NEGATIVE_FILE = DATA_DIR / "negative.csv"
POSITIVE_FILE = DATA_DIR / "positive.csv"

# Показываем папку, из которой сейчас работает Jupyter.
print("Рабочая папка:", Path.cwd())

# exists() проверяет, найден ли файл по указанному пути.
print("Найден negative.csv:", NEGATIVE_FILE.exists())
print("Найден positive.csv:", POSITIVE_FILE.exists())
```

```python
# ШАГ 02 — Загрузка CSV и перевод меток

# Читаем из каждого CSV два нужных столбца:
# ttext — текст сообщения, ttype — исходная метка (-1 или 1).
negative = pd.read_csv(
    NEGATIVE_FILE,
    usecols=["ttext", "ttype"],
)

positive = pd.read_csv(
    POSITIVE_FILE,
    usecols=["ttext", "ttype"],
)

# Задаём правило перевода меток:
# -1 — негатив — превращаем в 0;
#  1 — позитив — оставляем как 1.
label_mapping = {-1: 0, 1: 1}

# Создаём в каждой таблице новый столбец label.
# Исходный столбец ttype при этом сохраняется.
negative["label"] = negative["ttype"].map(label_mapping)
positive["label"] = positive["ttype"].map(label_mapping)

# Проверяем количество строк с каждой новой меткой.
print("Метки в negative.csv:")
print(negative["label"].value_counts())

print("\nМетки в positive.csv:")
print(positive["label"].value_counts())

# Показываем первые три строки каждой таблицы.
# Так можно сравнить исходную метку ttype с новой label.
display(negative[["ttext", "ttype", "label"]].head(3))
display(positive[["ttext", "ttype", "label"]].head(3))
```

```python
# ШАГ 03 — Объединение данных

# Ставим строки negative и positive друг под другом.
# В каждой строке сохраняются текст, исходная метка и новая метка label.
df = pd.concat([negative, positive], ignore_index=True)

print("Строк после объединения:", len(df))
print("Сообщений каждого класса:")
print(df["label"].value_counts().sort_index())
```

```python
# ШАГ 04 — Функция проверки и удаления повторов

def remove_text_duplicates(data, text_column):
    print(f"\n=== Проверка столбца {text_column} ===")

    # Создаём копию таблицы без отсутствующих значений
    # в проверяемом столбце.
    working = data.dropna(subset=[text_column]).copy()
    missing_count = len(data) - len(working)

    # Находим строки, в которых текст пустой или состоит из пробелов.
    blank = working[text_column].astype(str).str.strip().eq("")
    blank_count = int(blank.sum())

    # Удаляем такие строки.
    working = working.loc[~blank].copy()
    print("1. Строк без текста:", missing_count + blank_count)

    # Выбираем столбцы, которые покажем в таблицах с примерами.
    columns_to_show = [text_column, "label"]

    # При проверке очищенного текста показываем также исходный текст.
    if text_column != "ttext" and "ttext" in working.columns:
        columns_to_show = ["ttext", text_column, "label"]

    # Проверяем, есть ли одинаковые тексты с разными метками:
    # один и тот же текст размечен и как негатив, и как позитив.
    conflicting = (
        working.groupby(text_column)["label"].transform("nunique") > 1
    )

    # Получаем список разных текстов с противоречивыми метками.
    conflicting_texts = (
        working.loc[conflicting, text_column].drop_duplicates()
    )

    print(
        "2. Разных текстов с противоречивыми метками:",
        len(conflicting_texts),
    )

    if len(conflicting_texts) > 0:
        print("Примеры конфликтов — все эти строки будут удалены:")

        # Берём первые три конфликтующих текста.
        examples = conflicting_texts.head(3)

        # Показываем все строки с этими текстами и их метки.
        display(
            working.loc[
                working[text_column].isin(examples),
                columns_to_show,
            ].sort_values([text_column, "label"])
        )

    # Удаляем все строки с противоречивыми метками.
    working = working.loc[~conflicting].copy()

    # Для каждой строки считаем, сколько всего строк
    # имеют такой же текст в проверяемом столбце.
    repeat_count = (
        working.groupby(text_column)[text_column].transform("size")
    )

    # Первую строку с данным текстом оставляем.
    # Все последующие одинаковые строки помечаем как повторы.
    duplicated = working.duplicated(
        subset=[text_column],
        keep="first",
    )
    duplicate_count = int(duplicated.sum())

    print("3. Всего повторных строк для удаления:", duplicate_count)

    if duplicate_count > 0:
        # Берём строки, которые будут удалены как повторы.
        duplicate_examples = (
            working.loc[duplicated, columns_to_show].copy()
        )

        # Если текст встретился 4 раза, одну строку оставляем,
        # а 3 копии удаляем. Это число добавляем в таблицу.
        duplicate_examples["удаляется_копий"] = (
            repeat_count.loc[duplicated].to_numpy() - 1
        )

        # Каждый повторяющийся текст показываем только один раз.
        # Выводим пять текстов с наибольшим числом удаляемых копий.
        duplicate_examples = (
            duplicate_examples
            .drop_duplicates(subset=[text_column])
            .sort_values("удаляется_копий", ascending=False)
            .head(5)
        )

        print(
            "Примеры повторяющихся текстов "
            "и число удаляемых копий:"
        )
        display(duplicate_examples)

    # Удаляем все строки, помеченные как повторы.
    working = working.loc[~duplicated].copy()

    print("Итог: строк осталось:", len(working))

    # Возвращаем обработанную таблицу.
    return working
```

```python
# ШАГ 05 — Применение функции к исходному тексту

# Передаём функции общую таблицу df и имя столбца ttext.
# Функция покажет результаты проверки и вернёт таблицу без повторов.
df = remove_text_duplicates(df, "ttext")
```

```python
# ШАГ 06 — Функция очистки текста

# re — стандартная библиотека Python для поиска и замены по шаблонам.
import re

def clean_text(text):
    # Превращаем входное значение в строку.
    # lower() переводит заглавные буквы в строчные.
    text = str(text).lower()

    # Удаляем ссылки, начинающиеся с http://, https:// или www.
    # Вместо ссылки ставим пробел, чтобы соседние слова не склеились.
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    # Удаляем упоминания пользователей вместе с символом @.
    text = re.sub(r"@\w+", " ", text)

    # Удаляем хештеги целиком: и символ #, и слово после него.
    text = re.sub(r"#\w+", " ", text)

    # Оставляем русские буквы, цифры и пробелы; остальные символы заменяем пробелом.
    text = re.sub(r"[^а-яё0-9\s]", " ", text)

    # Заменяем несколько пробелов подряд одним пробелом.
    # strip() убирает пробелы в начале и конце строки.
    text = re.sub(r"\s+", " ", text).strip()

    # Отдаём готовый текст туда, откуда вызвали функцию.
    return text
```

```python
# ШАГ 07 — Применение очистки ко всем сообщениям

# Берём столбец ttext с исходными сообщениями.
# map(clean_text) вызывает нашу функцию clean_text для каждого сообщения.
# Результаты записываем в новый столбец text_clean.
# Исходный столбец ttext сохраняется: он пригодится для сравнения.
df["text_clean"] = df["ttext"].map(clean_text)

# Проверяем, сколько сообщений находится в таблице.
print("Всего сообщений:", len(df))

# После удаления ссылок, упоминаний и латиницы некоторые тексты
# могли стать пустыми. Считаем такие строки, но пока не удаляем.
print("Пустых текстов после очистки:", df["text_clean"].eq("").sum())

# Показываем пять примеров: исходный текст, очищенный текст и метку.
# random_state=42 позволяет при повторном запуске видеть те же примеры.
display(
    df[["ttext", "text_clean", "label"]].sample(5, random_state=42)
)

# Ищем сообщения, в которых есть хотя бы одно слово или число
# длиной от двух символов — такие признаки сможет взять TF-IDF.
has_usable_token = df["text_clean"].str.contains(
    r"\b[а-яё0-9]{2,}\b",
    regex=True,
)

print("Текстов без пригодных для TF-IDF слов:", (~has_usable_token).sum())

# Показываем примеры, включая исходный текст.
display(
    df.loc[~has_usable_token, ["ttext", "text_clean", "label"]].head(10)
)
```

```python
# ШАГ 08 — Удаление непригодных очищенных текстов

# Запоминаем число строк до удаления, чтобы оценить результат.
rows_before = len(df)

# Стандартный TF-IDF берёт слова длиной от двух символов.
# Если в text_clean нет ни одного такого слова или числа,
# модель не получит из этого сообщения полезных признаков.
# na=False считает отсутствующее значение непригодным.
has_usable_token = df["text_clean"].str.contains(
    r"\b[а-яё0-9]{2,}\b",
    regex=True,
    na=False,
)

# Оставляем только строки, в которых найден подходящий токен.
# copy() создаёт отдельную таблицу с оставшимися строками.
df = df.loc[has_usable_token].copy()

print("Строк до проверки:", rows_before)
print("Удалено строк без пригодных токенов:", rows_before - len(df))
print("Строк после проверки:", len(df))

# Теперь проверяем повторы уже по очищенному тексту.
# Функция также уберёт тексты с противоречивыми метками.
df = remove_text_duplicates(df, "text_clean")
```

```python
# ШАГ 09 — Разделение данных на обучение и тест

# Импортируем функцию, которая разделяет строки таблицы на две части.
from sklearn.model_selection import train_test_split

# Делим всю таблицу df.
# Текст и его метка остаются вместе в одной строке.
train_df, test_df = train_test_split(
    df,
    test_size=0.20,         # 20% строк — для теста, 80% — для обучения.
    stratify=df["label"],  # Сохраняем долю негатива и позитива в обеих частях.
    random_state=42,       # При повторном запуске получим то же разделение.
)

# Начинаем нумерацию строк в каждой новой таблице с нуля.
# drop=True означает, что старые номера строк не станут новым столбцом.
train_df = train_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

# Проверяем размер обеих частей и число сообщений каждого класса.
print("Обучающая выборка:", len(train_df), "строк")
print("Негатив (0) и позитив (1) в обучении:")
print(train_df["label"].value_counts().sort_index())

print("\nТестовая выборка:", len(test_df), "строк")
print("Негатив (0) и позитив (1) в тесте:")
print(test_df["label"].value_counts().sort_index())
```

```python
# ШАГ 10 — Словарь и TF-IDF-признаки для обучения

# Импортируем инструмент, который превращает тексты в числовые признаки.
from sklearn.feature_extraction.text import TfidfVectorizer

# Задаём правила построения словаря.
# На этой строке словарь ещё не создаётся.
vectorizer = TfidfVectorizer(
    max_features=None,    # Не задаём предельное число слов заранее.
    min_df=3,             # Слово должно быть минимум в 3 обучающих сообщениях.
    max_df=0.95,          # Исключаем слова из более чем 95% сообщений.
    ngram_range=(1, 1),   # Каждый признак — одно слово.
)

# Строим словарь ТОЛЬКО по обучающим текстам.
# fit изучает слова и рассчитывает веса IDF.
# transform превращает каждый текст в строку числовой матрицы.
X_train_tfidf = vectorizer.fit_transform(
    train_df["text_clean"]
)

# Получаем названия признаков — слова из построенного словаря.
# Они понадобятся позже для анализа весов MLP.
feature_names = vectorizer.get_feature_names_out()

# Проверяем результат: строки — сообщения, столбцы — признаки.
print("Обучающих сообщений:", X_train_tfidf.shape[0])
print("Слов-признаков в словаре(столько столбцов в матрице):", X_train_tfidf.shape[1])
print("Примеры признаков:", list(feature_names[:15]))
# Отбираем первые 15 признаков, начинающихся с буквы.
russian_examples = [
    word for word in feature_names
    if word[0].isalpha()
][:15]

print("Примеры слов:", russian_examples)
```

```python
# ШАГ 11 — Просмотр небольшого фрагмента матрицы TF-IDF

# Выберем первые три обучающих сообщения.
rows_to_show = 3
small_matrix = X_train_tfidf[:rows_to_show]

# Для наглядности выберем 10 признаков с наибольшими весами
# именно в этих трёх сообщениях.
# Это только выбор столбцов для показа: словарь и модель не меняются.
feature_scores = small_matrix.sum(axis=0).A1
column_indices = feature_scores.argsort()[::-1][:10]

# Показываем тексты и правильные метки этих строк.
print("Сообщения, которым соответствуют строки матрицы:")
display(train_df[["text_clean", "label"]].head(rows_to_show))

# Берём только 3 строки и 10 столбцов.
# toarray() здесь безопасен: превращаем в обычную таблицу
# лишь маленький фрагмент, а не всю большую матрицу.
preview = pd.DataFrame(
    small_matrix[:, column_indices].toarray(),
    columns=feature_names[column_indices],
    index=train_df.index[:rows_to_show],
)

print("Фрагмент матрицы TF-IDF:")
display(preview.round(3))
```

```python


# ШАГ 12 — Обучение классификатора MLP

# 1. Импортируем готовую реализацию многослойного перцептрона.
from sklearn.neural_network import MLPClassifier


# 2. Подготавливаем правильные ответы для обучающих сообщений.
# В X_train_tfidf находятся TF-IDF-признаки сообщений.
# В столбце label находятся соответствующие им правильные метки:
# 0 — негатив, 1 — позитив.
# Порядок строк в X_train_tfidf и train_df одинаковый.
y_train = train_df["label"]

print("Сообщений для обучения:", X_train_tfidf.shape[0])
print("Признаков у каждого сообщения:", X_train_tfidf.shape[1])
print("Правильных ответов:", len(y_train))


# 3. Создаём модель и задаём её параметры.
# На этом этапе модель ещё не обучена.
mlp = MLPClassifier(
    # Один скрытый слой из 32 нейронов.
    # Каждый нейрон получает признаки сообщения и имеет свои веса.
    hidden_layer_sizes=(32,),

    # ReLU преобразует результат каждого скрытого нейрона:
    # отрицательные значения превращаются в 0,
    # положительные остаются положительными.
    activation="relu",

    # Adam — алгоритм, который изменяет веса модели
    # на основе ошибок в её предсказаниях.
    solver="adam",

    # Не более 20 эпох — проходов по обучающим сообщениям.
    max_iter=20,

    # Отложить часть обучающих данных для внутренней проверки.
    # Если качество перестанет улучшаться, закончить раньше.
    early_stopping=True,

    # Для внутренней проверки используется 10% train_df.
    # Наша отдельная тестовая выборка здесь не используется.
    validation_fraction=0.10,

    # Если три эпохи подряд нет достаточного улучшения
    # на внутренней проверке, остановить обучение.
    n_iter_no_change=3,

    # Фиксируем случайность при создании начальных весов
    # и внутреннем разделении данных.
    random_state=42,

    # После каждой эпохи показываем ход обучения в Jupyter.
    verbose=True,
)


# 4. Запускаем обучение.
# Для каждого сообщения модель получает строку TF-IDF-признаков
# и сравнивает свой ответ с правильной меткой из y_train.
# Меняются веса MLP; матрица X_train_tfidf не перезаписывается.
mlp.fit(X_train_tfidf, y_train)


# 5. Смотрим, что получилось после обучения.
print("\nОбучение завершено.")
print("Выполнено эпох:", mlp.n_iter_)

# Первая матрица весов соединяет входные признаки
# с 32 скрытыми нейронами: (число признаков, 32).
print("Форма весов первого слоя:", mlp.coefs_[0].shape)

# Вторая матрица соединяет 32 скрытых нейрона
# с одним выходом для бинарной классификации: (32, 1).
print("Форма весов выходного слоя:", mlp.coefs_[1].shape)

# Это значение одного обученного веса, а не вероятность.
# Вес может быть отрицательным или больше единицы.
print("Пример одного веса:", round(mlp.coefs_[0][0, 0], 4))

# Accuracy на части обучающих данных, автоматически отложенной
# для ранней остановки. Это ещё не результат на тестовой выборке.
print(
    "Лучшая accuracy на внутренней проверке:",
    round(mlp.best_validation_score_, 4),
)
```

```python
# ШАГ 13 — Оценка модели на тестовой выборке

# Импортируем функции для расчёта качества и построения матрицы ошибок.
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
)

# Библиотека для отображения графика.
import matplotlib.pyplot as plt


# 1. Превращаем тестовые тексты в числовые признаки.
# Используем ТОЛЬКО transform: словарь и веса IDF уже были
# построены на обучающей выборке в шаге 10.
X_test_tfidf = vectorizer.transform(test_df["text_clean"])

# Настоящие метки тестовых сообщений хранятся отдельно.
y_test = test_df["label"]

print("Тестовых сообщений:", X_test_tfidf.shape[0])
print("Признаков у сообщения:", X_test_tfidf.shape[1])


# 2. Просим обученную MLP определить тональность тестовых сообщений.
# Модель возвращает массив предсказанных меток 0 или 1.
y_pred = mlp.predict(X_test_tfidf)


# 3. Считаем общую долю правильных ответов по обоим классам.
accuracy = accuracy_score(y_test, y_pred)
print("\nAccuracy по всем тестовым сообщениям:", round(accuracy, 4))


# 4. Считаем Precision, Recall и F1 отдельно для негатива и позитива.
# pos_label указывает, какой класс сейчас считаем интересующим нас.
metric_rows = []

for class_id, class_name in [(0, "Негатив"), (1, "Позитив")]:
    metric_rows.append({
        "Класс": f"{class_name} ({class_id})",
        "Сообщений в тесте": int((y_test == class_id).sum()),
        "Precision": precision_score(
            y_test, y_pred, pos_label=class_id, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, pos_label=class_id, zero_division=0
        ),
        "F1": f1_score(
            y_test, y_pred, pos_label=class_id, zero_division=0
        ),
    })

metrics_table = pd.DataFrame(metric_rows).set_index("Класс")

print("\nМетрики для каждого класса:")
display(metrics_table.round(4))


# 5. Строим матрицу ошибок.
# Строки — настоящие метки; столбцы — предсказания модели.
# Порядок классов явно задаём как 0, затем 1.
cm = confusion_matrix(y_test, y_pred, labels=[0, 1])

print("Всего сообщений в матрице ошибок:", cm.sum())
print("Негатив ошибочно назван позитивом (FP):", cm[0, 1])
print("Позитив ошибочно назван негативом (FN):", cm[1, 0])

display_plot = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Негатив (0)", "Позитив (1)"],
)
display_plot.plot(cmap="Blues", values_format="d")
plt.title("Матрица ошибок на тестовой выборке")
plt.tight_layout()
plt.show()
```

```python
# ШАГ 14 — Просмотр ошибок модели на тестовых сообщениях

# Создаём таблицу для анализа.
# copy() сохраняет исходную test_df без изменений.
test_results = test_df[["ttext", "text_clean", "label"]].copy()

# Добавляем ответ модели из шага 13.
# Порядок y_pred совпадает с порядком строк test_df.
test_results["predicted_label"] = y_pred

# Узнаём, какому столбцу predict_proba соответствует класс 1.
# Это надёжнее, чем предполагать номер столбца.
positive_column = list(mlp.classes_).index(1)

# predict_proba возвращает вероятности классов для каждого сообщения.
# Берём вероятность позитива (класса 1).
test_results["p_positive"] = mlp.predict_proba(
    X_test_tfidf
)[:, positive_column]

# Группа FP: настоящий негатив (0) ошибочно назван позитивом (1).
false_positive = test_results.loc[
    (test_results["label"] == 0)
    & (test_results["predicted_label"] == 1)
]

# Группа FN: настоящий позитив (1) ошибочно назван негативом (0).
false_negative = test_results.loc[
    (test_results["label"] == 1)
    & (test_results["predicted_label"] == 0)
]

print("Негатив ошибочно назван позитивом (FP):", len(false_positive))
print("Позитив ошибочно назван негативом (FN):", len(false_negative))

# Столбцы, которые хотим видеть в примерах.
columns_to_show = [
    "ttext",
    "text_clean",
    "label",
    "predicted_label",
    "p_positive",
]

# Временно разрешаем Jupyter показывать длинные тексты целиком.
with pd.option_context("display.max_colwidth", None):
    print("\nПримеры FP — модель ошибочно увидела позитив:")
    display(
        false_positive[columns_to_show]
        .sample(n=min(5, len(false_positive)), random_state=42)
        .round({"p_positive": 3})
    )

    print("\nПримеры FN — модель ошибочно увидела негатив:")
    display(
        false_negative[columns_to_show]
        .sample(n=min(5, len(false_negative)), random_state=42)
        .round({"p_positive": 3})
    )
```

```python
# ШАГ 15 — Обучение второй модели: 64 нейрона

from sklearn.neural_network import MLPClassifier

# Создаём НОВУЮ модель.
# Старую модель mlp с 32 нейронами не меняем.
mlp_64 = MLPClassifier(
    hidden_layer_sizes=(64,),  # Один скрытый слой из 64 нейронов.
    activation="relu",         # Та же функция активации, что и раньше.
    solver="adam",             # Тот же алгоритм изменения весов.
    max_iter=20,               # Не более 20 эпох.
    early_stopping=True,       # Остановка при отсутствии улучшения.
    validation_fraction=0.10,  # Внутренняя проверка на 10% обучения.
    n_iter_no_change=3,        # Остановка после 3 эпох без улучшения.
    random_state=42,           # Фиксируем случайность.
    verbose=True,              # Показываем ход каждой эпохи.
)

print("Обучающих сообщений:", X_train_tfidf.shape[0])
print("Признаков у сообщения:", X_train_tfidf.shape[1])
print("Правильных меток:", len(y_train))

# Используем ТОЧНО ТУ ЖЕ матрицу и те же метки,
# которые передавали модели с 32 нейронами на шаге 12.
mlp_64.fit(X_train_tfidf, y_train)

print("\nОбучение модели с 64 нейронами завершено.")
print("Выполнено эпох:", mlp_64.n_iter_)
print("Форма весов первого слоя:", mlp_64.coefs_[0].shape)
print("Форма весов выходного слоя:", mlp_64.coefs_[1].shape)
print(
    "Лучшая accuracy на внутренней проверке:",
    round(mlp_64.best_validation_score_, 4),
)
```

```python
# ШАГ 16 — Сравнение моделей с 32 и 64 нейронами

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
)

# Для модели с 32 нейронами ответы y_pred уже получены на шаге 13.
# Получаем ответы новой модели на ТОЙ ЖЕ тестовой матрице.
y_pred_64 = mlp_64.predict(X_test_tfidf)

# Собираем модели и их ответы в один список.
# Это позволит рассчитать метрики одинаковым кодом.
model_predictions = [
    ("32 нейрона", y_pred),
    ("64 нейрона", y_pred_64),
]

# Здесь будем хранить общие результаты двух моделей.
overall_rows = []

# Здесь будем хранить Precision, Recall и F1 для каждого класса.
class_rows = []

# Здесь сохраним матрицы ошибок для построения графиков.
confusion_matrices = {}

for model_name, predicted_labels in model_predictions:
    # Доля правильных ответов среди всех тестовых сообщений.
    model_accuracy = accuracy_score(y_test, predicted_labels)

    # Матрица ошибок:
    # строки — настоящие классы, столбцы — ответы модели.
    model_cm = confusion_matrix(
        y_test,
        predicted_labels,
        labels=[0, 1],
    )

    confusion_matrices[model_name] = model_cm

    # Общие числа для сравнения.
    overall_rows.append({
        "Модель": model_name,
        "Accuracy": model_accuracy,
        "Правильных ответов": int(model_cm[0, 0] + model_cm[1, 1]),
        "Ошибок FP": int(model_cm[0, 1]),
        "Ошибок FN": int(model_cm[1, 0]),
    })

    # Оцениваем негатив и позитив по отдельности.
    for class_id, class_name in [(0, "Негатив"), (1, "Позитив")]:
        class_rows.append({
            "Модель": model_name,
            "Класс": f"{class_name} ({class_id})",
            "Сообщений в тесте": int((y_test == class_id).sum()),
            "Precision": precision_score(
                y_test,
                predicted_labels,
                pos_label=class_id,
                zero_division=0,
            ),
            "Recall": recall_score(
                y_test,
                predicted_labels,
                pos_label=class_id,
                zero_division=0,
            ),
            "F1": f1_score(
                y_test,
                predicted_labels,
                pos_label=class_id,
                zero_division=0,
            ),
        })

# Превращаем результаты в таблицы.
overall_comparison = pd.DataFrame(overall_rows).set_index("Модель")
class_comparison = pd.DataFrame(class_rows).set_index(
    ["Модель", "Класс"]
)

print("Тестовых сообщений:", len(y_test))

print("\nОбщее сравнение:")
display(overall_comparison.round(4))

print("\nМетрики отдельно для негатива и позитива:")
display(class_comparison.round(4))

# Показываем матрицы ошибок рядом.
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for ax, (model_name, _) in zip(axes, model_predictions):
    ConfusionMatrixDisplay(
        confusion_matrix=confusion_matrices[model_name],
        display_labels=["Негатив (0)", "Позитив (1)"],
    ).plot(
        ax=ax,
        cmap="Blues",
        values_format="d",
        colorbar=False,
    )

    ax.set_title(model_name)

fig.suptitle("Матрицы ошибок на одной тестовой выборке")
fig.tight_layout()
plt.show()
```

```python
# ШАГ 17 — Смотрим, какие ошибки делает модель с 64 нейронами

# Берём исходный текст, очищенный текст и правильную метку.
# copy() оставляет test_df без изменений.
comparison_examples = test_df[
    ["ttext", "text_clean", "label"]
].copy()

# Добавляем ответы обеих моделей.
comparison_examples["predicted_32"] = y_pred
comparison_examples["predicted_64"] = y_pred_64

# Находим столбец вероятности позитива у каждой модели.
positive_column_32 = list(mlp.classes_).index(1)
positive_column_64 = list(mlp_64.classes_).index(1)

# Добавляем вероятность позитива от каждой модели.
comparison_examples["p_positive_32"] = mlp.predict_proba(
    X_test_tfidf
)[:, positive_column_32]

comparison_examples["p_positive_64"] = mlp_64.predict_proba(
    X_test_tfidf
)[:, positive_column_64]

# FP модели с 64 нейронами:
# настоящий негатив (0) ошибочно назван позитивом (1).
false_positive_64 = comparison_examples.loc[
    (comparison_examples["label"] == 0)
    & (comparison_examples["predicted_64"] == 1)
]

# FN модели с 64 нейронами:
# настоящий позитив (1) ошибочно назван негативом (0).
false_negative_64 = comparison_examples.loc[
    (comparison_examples["label"] == 1)
    & (comparison_examples["predicted_64"] == 0)
]

# Сообщения, которые старая модель определяла неправильно,
# а новая определила правильно.
fixed_by_64 = comparison_examples.loc[
    (comparison_examples["predicted_32"] != comparison_examples["label"])
    & (comparison_examples["predicted_64"] == comparison_examples["label"])
]

# Сообщения, которые старая модель определяла правильно,
# а новая стала определять неправильно.
broken_by_64 = comparison_examples.loc[
    (comparison_examples["predicted_32"] == comparison_examples["label"])
    & (comparison_examples["predicted_64"] != comparison_examples["label"])
]

print("FP модели с 64 нейронами:", len(false_positive_64))
print("FN модели с 64 нейронами:", len(false_negative_64))
print("Ошибки 32 нейронов, исправленные 64 нейронами:", len(fixed_by_64))
print("Новые ошибки модели с 64 нейронами:", len(broken_by_64))

# Эти столбцы показывают один и тот же текст,
# ответы обеих моделей и вероятности позитива.
columns_to_compare = [
    "ttext",
    "text_clean",
    "label",
    "predicted_32",
    "predicted_64",
    "p_positive_32",
    "p_positive_64",
]

# Временно показываем длинные тексты целиком.
with pd.option_context("display.max_colwidth", None):
    print("\nПримеры FP модели с 64 нейронами:")
    display(
        false_positive_64[columns_to_compare]
        .sample(n=min(5, len(false_positive_64)), random_state=42)
        .round(3)
    )

    print("\nПримеры FN модели с 64 нейронами:")
    display(
        false_negative_64[columns_to_compare]
        .sample(n=min(5, len(false_negative_64)), random_state=42)
        .round(3)
    )

    print("\nПримеры ошибок, исправленных моделью с 64 нейронами:")
    display(
        fixed_by_64[columns_to_compare]
        .sample(n=min(5, len(fixed_by_64)), random_state=42)
        .round(3)
    )

    print("\nПримеры новых ошибок модели с 64 нейронами:")
    display(
        broken_by_64[columns_to_compare]
        .sample(n=min(5, len(broken_by_64)), random_state=42)
        .round(3)
    )
```

```python
# ШАГ 18 — Сколько предсказаний находится рядом с порогом 0,5

import numpy as np
import pandas as pd

# Обе модели проверяем на одной и той же тестовой матрице.
models_for_threshold = [
    ("32 нейрона", mlp, y_pred),
    ("64 нейрона", mlp_64, y_pred_64),
]

# Сюда добавим результаты подсчёта для каждой модели.
threshold_rows = []

for model_name, model, predicted_labels in models_for_threshold:
    # Находим столбец вероятности позитива — класса 1.
    positive_column = list(model.classes_).index(1)

    # Получаем вероятность позитива для каждого тестового сообщения.
    probabilities = model.predict_proba(
        X_test_tfidf
    )[:, positive_column]

    # Вычисляем расстояние каждой вероятности до порога 0,5.
    # Например, для 0,503 расстояние равно 0,003.
    distance_to_half = np.abs(probabilities - 0.5)

    # Сообщения с вероятностью от 0,49 до 0,51 включительно.
    near_001 = distance_to_half <= 0.01

    # Сообщения с вероятностью от 0,45 до 0,55 включительно.
    near_005 = distance_to_half <= 0.05

    # Считаем ошибочные ответы среди сообщений около порога.
    wrong_near_005 = near_005 & (
        predicted_labels != y_test.to_numpy()
    )

    threshold_rows.append({
        "Модель": model_name,
        "Ровно 0,5": int(np.count_nonzero(probabilities == 0.5)),
        "Выглядит как 0,500 после округления": int(
            np.count_nonzero(np.round(probabilities, 3) == 0.5)
        ),
        "От 0,49 до 0,51": int(np.count_nonzero(near_001)),
        "От 0,45 до 0,55": int(np.count_nonzero(near_005)),
        "Ошибок в диапазоне 0,45–0,55": int(
            np.count_nonzero(wrong_near_005)
        ),
    })

# Показываем результаты в одной таблице.
threshold_table = pd.DataFrame(threshold_rows).set_index("Модель")

print("Всего тестовых сообщений:", len(y_test))
display(threshold_table)
```
