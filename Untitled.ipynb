# Классификатор тональности

Цель: научить модель различать позитивные и негативные сообщения.

## 1. Подготовка библиотек

```python
# ШАГ 01 — Пути к файлам

# Path — инструмент Python для работы с путями к папкам и файлам.
from pathlib import Path

# Pandas понадобится для чтения CSV и работы с таблицами.
# Имя pd — принятое короткое обозначение pandas.
import pandas as pd

# Указываем папку, в которой лежат исходные данные.
DATA_DIR = Path("data")

# Составляем путь к каждому файлу.
# Знак / здесь соединяет путь к папке с именем файла.
NEGATIVE_FILE = DATA_DIR / "negative.csv"
POSITIVE_FILE = DATA_DIR / "positive.csv"

# Показываем папку, из которой сейчас работает Jupyter.
print("Рабочая папка:", Path.cwd())

# exists() проверяет, найден ли файл по указанному пути.
print("Найден negative.csv:", NEGATIVE_FILE.exists())
print("Найден positive.csv:", POSITIVE_FILE.exists())
```

```python
# ШАГ 02 — Загрузка CSV и перевод меток

# Читаем из каждого CSV два нужных столбца:
# ttext — текст сообщения, ttype — исходная метка (-1 или 1).
negative = pd.read_csv(
    NEGATIVE_FILE,
    usecols=["ttext", "ttype"],
)

positive = pd.read_csv(
    POSITIVE_FILE,
    usecols=["ttext", "ttype"],
)

# Задаём правило перевода меток:
# -1 — негатив — превращаем в 0;
#  1 — позитив — оставляем как 1.
label_mapping = {-1: 0, 1: 1}

# Создаём в каждой таблице новый столбец label.
# Исходный столбец ttype при этом сохраняется.
negative["label"] = negative["ttype"].map(label_mapping)
positive["label"] = positive["ttype"].map(label_mapping)

# Проверяем количество строк с каждой новой меткой.
print("Метки в negative.csv:")
print(negative["label"].value_counts())

print("\nМетки в positive.csv:")
print(positive["label"].value_counts())

# Показываем первые три строки каждой таблицы.
# Так можно сравнить исходную метку ttype с новой label.
display(negative[["ttext", "ttype", "label"]].head(3))
display(positive[["ttext", "ttype", "label"]].head(3))
```

```python
# ШАГ 03 — Объединение данных

# Ставим строки negative и positive друг под другом.
# В каждой строке сохраняются текст, исходная метка и новая метка label.
df = pd.concat([negative, positive], ignore_index=True)

print("Строк после объединения:", len(df))
print("Сообщений каждого класса:")
print(df["label"].value_counts().sort_index())
```

```python
# ШАГ 04 — Функция проверки и удаления повторов

def remove_text_duplicates(data, text_column):
    print(f"\n=== Проверка столбца {text_column} ===")

    # Создаём копию таблицы без отсутствующих значений
    # в проверяемом столбце.
    working = data.dropna(subset=[text_column]).copy()
    missing_count = len(data) - len(working)

    # Находим строки, в которых текст пустой или состоит из пробелов.
    blank = working[text_column].astype(str).str.strip().eq("")
    blank_count = int(blank.sum())

    # Удаляем такие строки.
    working = working.loc[~blank].copy()
    print("1. Строк без текста:", missing_count + blank_count)

    # Выбираем столбцы, которые покажем в таблицах с примерами.
    columns_to_show = [text_column, "label"]

    # При проверке очищенного текста показываем также исходный текст.
    if text_column != "ttext" and "ttext" in working.columns:
        columns_to_show = ["ttext", text_column, "label"]

    # Проверяем, есть ли одинаковые тексты с разными метками:
    # один и тот же текст размечен и как негатив, и как позитив.
    conflicting = (
        working.groupby(text_column)["label"].transform("nunique") > 1
    )

    # Получаем список разных текстов с противоречивыми метками.
    conflicting_texts = (
        working.loc[conflicting, text_column].drop_duplicates()
    )

    print(
        "2. Разных текстов с противоречивыми метками:",
        len(conflicting_texts),
    )

    if len(conflicting_texts) > 0:
        print("Примеры конфликтов — все эти строки будут удалены:")

        # Берём первые три конфликтующих текста.
        examples = conflicting_texts.head(3)

        # Показываем все строки с этими текстами и их метки.
        display(
            working.loc[
                working[text_column].isin(examples),
                columns_to_show,
            ].sort_values([text_column, "label"])
        )

    # Удаляем все строки с противоречивыми метками.
    working = working.loc[~conflicting].copy()

    # Для каждой строки считаем, сколько всего строк
    # имеют такой же текст в проверяемом столбце.
    repeat_count = (
        working.groupby(text_column)[text_column].transform("size")
    )

    # Первую строку с данным текстом оставляем.
    # Все последующие одинаковые строки помечаем как повторы.
    duplicated = working.duplicated(
        subset=[text_column],
        keep="first",
    )
    duplicate_count = int(duplicated.sum())

    print("3. Всего повторных строк для удаления:", duplicate_count)

    if duplicate_count > 0:
        # Берём строки, которые будут удалены как повторы.
        duplicate_examples = (
            working.loc[duplicated, columns_to_show].copy()
        )

        # Если текст встретился 4 раза, одну строку оставляем,
        # а 3 копии удаляем. Это число добавляем в таблицу.
        duplicate_examples["удаляется_копий"] = (
            repeat_count.loc[duplicated].to_numpy() - 1
        )

        # Каждый повторяющийся текст показываем только один раз.
        # Выводим пять текстов с наибольшим числом удаляемых копий.
        duplicate_examples = (
            duplicate_examples
            .drop_duplicates(subset=[text_column])
            .sort_values("удаляется_копий", ascending=False)
            .head(5)
        )

        print(
            "Примеры повторяющихся текстов "
            "и число удаляемых копий:"
        )
        display(duplicate_examples)

    # Удаляем все строки, помеченные как повторы.
    working = working.loc[~duplicated].copy()

    print("Итог: строк осталось:", len(working))

    # Возвращаем обработанную таблицу.
    return working
```

```python
# ШАГ 05 — Применение функции к исходному тексту

# Передаём функции общую таблицу df и имя столбца ttext.
# Функция покажет результаты проверки и вернёт таблицу без повторов.
df = remove_text_duplicates(df, "ttext")
```

```python
# ШАГ 06 — Функция очистки текста

# re — стандартная библиотека Python для поиска и замены по шаблонам.
import re

def clean_text(text):
    # Превращаем входное значение в строку.
    # lower() переводит заглавные буквы в строчные.
    text = str(text).lower()

    # Удаляем ссылки, начинающиеся с http://, https:// или www.
    # Вместо ссылки ставим пробел, чтобы соседние слова не склеились.
    text = re.sub(r"https?://[^\s()]+|www\.[^\s()]+", " ", text)

    # Удаляем упоминания пользователей вместе с символом @.
    text = re.sub(r"@\w+", " ", text)

    # Удаляем хештеги целиком: и символ #, и слово после него.
    text = re.sub(r"#\w+", " ", text)

    # Оставляем русские буквы, цифры, пробелы и скобки.
    text = re.sub(r"[^а-яё0-9\s()]", " ", text)

    # Отделяем серии скобок от слов: спасибо))) -> спасибо ))).
    text = re.sub(r"(\){2,}|\({2,})", r" \1 ", text)

    # Одиночные скобки не считаем смайлами и удаляем.
    text = re.sub(r"(?<!\()\((?!\()|(?<!\))\)(?!\))", " ", text)

    # Заменяем несколько пробелов подряд одним пробелом.
    # strip() убирает пробелы в начале и конце строки.
    text = re.sub(r"\s+", " ", text).strip()

    # Отдаём готовый текст туда, откуда вызвали функцию.
    return text

# Проверка сразу после запуска этой ячейки.
# Если скобки исчезли, выполнение остановится с AssertionError.
assert clean_text("Отлично))))") == "отлично ))))"
assert clean_text("Ужас(((((") == "ужас ((((("
print("Проверка очистки: Отлично)))) →", clean_text("Отлично))))"))
print("Проверка очистки: Ужас((((( →", clean_text("Ужас((((("))
```

```python
# ШАГ 07 — Применение очистки ко всем сообщениям

# Берём столбец ttext с исходными сообщениями.
# map(clean_text) вызывает нашу функцию clean_text для каждого сообщения.
# Результаты записываем в новый столбец text_clean.
# Исходный столбец ttext сохраняется: он пригодится для сравнения.
df["text_clean"] = df["ttext"].map(clean_text)

# Проверяем, сколько сообщений находится в таблице.
print("Всего сообщений:", len(df))

# После удаления ссылок, упоминаний и латиницы некоторые тексты
# могли стать пустыми. Считаем такие строки, но пока не удаляем.
print("Пустых текстов после очистки:", df["text_clean"].eq("").sum())

# Показываем пять примеров: исходный текст, очищенный текст и метку.
# random_state=42 позволяет при повторном запуске видеть те же примеры.
display(
    df[["ttext", "text_clean", "label"]].sample(5, random_state=42)
)

# Ищем слово/число длиной от двух символов или серию скобок.
# Такие признаки сможет взять TF-IDF на шаге 10.
has_usable_token = df["text_clean"].str.contains(
    r"\b[а-яё0-9]{2,}\b|\){2,}|\({2,}",
    regex=True,
)

print("Текстов без пригодных для TF-IDF токенов:", (~has_usable_token).sum())

# Показываем примеры, включая исходный текст.
display(
    df.loc[~has_usable_token, ["ttext", "text_clean", "label"]].head(10)
)
```

```python
# ШАГ 08 — Удаление непригодных очищенных текстов

# Запоминаем число строк до удаления, чтобы оценить результат.
rows_before = len(df)

# Наш TF-IDF берёт слова/числа длиной от двух символов
# и серии одинаковых скобок длиной от двух символов.
# Если в text_clean нет ни одного такого токена,
# модель не получит из этого сообщения полезных признаков.
# na=False считает отсутствующее значение непригодным.
has_usable_token = df["text_clean"].str.contains(
    r"\b[а-яё0-9]{2,}\b|\){2,}|\({2,}",
    regex=True,
    na=False,
)

# Оставляем только строки, в которых найден подходящий токен.
# copy() создаёт отдельную таблицу с оставшимися строками.
df = df.loc[has_usable_token].copy()

print("Строк до проверки:", rows_before)
print("Удалено строк без пригодных токенов:", rows_before - len(df))
print("Строк после проверки:", len(df))

# Теперь проверяем повторы уже по очищенному тексту.
# Функция также уберёт тексты с противоречивыми метками.
df = remove_text_duplicates(df, "text_clean")
```

```python
# ШАГ 09 — Разделение данных на обучение и тест

# Импортируем функцию, которая разделяет строки таблицы на две части.
from sklearn.model_selection import train_test_split

# Делим всю таблицу df.
# Текст и его метка остаются вместе в одной строке.
train_df, test_df = train_test_split(
    df,
    test_size=0.20,         # 20% строк — для теста, 80% — для обучения.
    stratify=df["label"],  # Сохраняем долю негатива и позитива в обеих частях.
    random_state=42,       # При повторном запуске получим то же разделение.
)

# Начинаем нумерацию строк в каждой новой таблице с нуля.
# drop=True означает, что старые номера строк не станут новым столбцом.
train_df = train_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

# Проверяем размер обеих частей и число сообщений каждого класса.
print("Обучающая выборка:", len(train_df), "строк")
print("Негатив (0) и позитив (1) в обучении:")
print(train_df["label"].value_counts().sort_index())

print("\nТестовая выборка:", len(test_df), "строк")
print("Негатив (0) и позитив (1) в тесте:")
print(test_df["label"].value_counts().sort_index())
```

```python
# ШАГ 10 — Словарь и TF-IDF-признаки для обучения

# Импортируем инструмент, который превращает тексты в числовые признаки.
from sklearn.feature_extraction.text import TfidfVectorizer

# Задаём правила построения словаря.
# На этой строке словарь ещё не создаётся.
vectorizer = TfidfVectorizer(
    max_features=None,    # Не задаём предельное число слов заранее.
    min_df=3,             # Слово должно быть минимум в 3 обучающих сообщениях.
    max_df=0.95,          # Исключаем слова из более чем 95% сообщений.
    ngram_range=(1, 1),   # Каждый признак — слово/число или смайл.
    # Без token_pattern TF-IDF проигнорировал бы скобки.
    token_pattern=r"(?u)\b[а-яё0-9]{2,}\b|\){2,}|\({2,}",
)

# Строим словарь ТОЛЬКО по обучающим текстам.
# fit изучает слова и рассчитывает веса IDF.
# transform превращает каждый текст в строку числовой матрицы.
X_train_tfidf = vectorizer.fit_transform(
    train_df["text_clean"]
)

# Получаем названия признаков — слова, числа и смайлы из словаря.
# Они понадобятся позже для анализа весов MLP.
feature_names = vectorizer.get_feature_names_out()

# Проверяем результат: строки — сообщения, столбцы — признаки.
print("Обучающих сообщений:", X_train_tfidf.shape[0])
print("Признаков в словаре (столбцов в матрице):", X_train_tfidf.shape[1])
print("Примеры признаков:", list(feature_names[:15]))
# Отбираем первые 15 признаков, начинающихся с буквы.
russian_examples = [
    word for word in feature_names
    if word[0].isalpha()
][:15]

print("Примеры слов:", russian_examples)

# Показываем смайлы, которые встретились минимум в min_df=3 текстах обучения.
smile_features = [token for token in feature_names if token.startswith(("(", ")"))]
print("Смайлов-признаков в словаре:", len(smile_features))
print("Примеры смайлов-признаков:", smile_features[:15])
```

```python
# ШАГ 11 — Просмотр небольшого фрагмента матрицы TF-IDF

# Выберем первые три обучающих сообщения.
rows_to_show = 3
small_matrix = X_train_tfidf[:rows_to_show]

# Для наглядности выберем 10 признаков с наибольшими весами
# именно в этих трёх сообщениях.
# Это только выбор столбцов для показа: словарь и модель не меняются.
feature_scores = small_matrix.sum(axis=0).A1
column_indices = feature_scores.argsort()[::-1][:10]

# Показываем тексты и правильные метки этих строк.
print("Сообщения, которым соответствуют строки матрицы:")
display(train_df[["text_clean", "label"]].head(rows_to_show))

# Берём только 3 строки и 10 столбцов.
# toarray() здесь безопасен: превращаем в обычную таблицу
# лишь маленький фрагмент, а не всю большую матрицу.
preview = pd.DataFrame(
    small_matrix[:, column_indices].toarray(),
    columns=feature_names[column_indices],
    index=train_df.index[:rows_to_show],
)

print("Фрагмент матрицы TF-IDF:")
display(preview.round(3))
```

```python


# ШАГ 12 — Обучение классификатора MLP

# 1. Импортируем готовую реализацию многослойного перцептрона.
from sklearn.neural_network import MLPClassifier


# 2. Подготавливаем правильные ответы для обучающих сообщений.
# В X_train_tfidf находятся TF-IDF-признаки сообщений.
# В столбце label находятся соответствующие им правильные метки:
# 0 — негатив, 1 — позитив.
# Порядок строк в X_train_tfidf и train_df одинаковый.
y_train = train_df["label"]

print("Сообщений для обучения:", X_train_tfidf.shape[0])
print("Признаков у каждого сообщения:", X_train_tfidf.shape[1])
print("Правильных ответов:", len(y_train))


# 3. Создаём модель и задаём её параметры.
# На этом этапе модель ещё не обучена.
mlp = MLPClassifier(
    # Один скрытый слой из 32 нейронов.
    # Каждый нейрон получает признаки сообщения и имеет свои веса.
    hidden_layer_sizes=(32,),

    # ReLU преобразует результат каждого скрытого нейрона:
    # отрицательные значения превращаются в 0,
    # положительные остаются положительными.
    activation="relu",

    # Adam — алгоритм, который изменяет веса модели
    # на основе ошибок в её предсказаниях.
    solver="adam",

    # Не более 20 эпох — проходов по обучающим сообщениям.
    max_iter=20,

    # Отложить часть обучающих данных для внутренней проверки.
    # Если качество перестанет улучшаться, закончить раньше.
    early_stopping=True,

    # Для внутренней проверки используется 10% train_df.
    # Наша отдельная тестовая выборка здесь не используется.
    validation_fraction=0.10,

    # Если три эпохи подряд нет достаточного улучшения
    # на внутренней проверке, остановить обучение.
    n_iter_no_change=3,

    # Фиксируем случайность при создании начальных весов
    # и внутреннем разделении данных.
    random_state=42,

    # После каждой эпохи показываем ход обучения в Jupyter.
    verbose=True,
)


# 4. Запускаем обучение.
# Для каждого сообщения модель получает строку TF-IDF-признаков
# и сравнивает свой ответ с правильной меткой из y_train.
# Меняются веса MLP; матрица X_train_tfidf не перезаписывается.
mlp.fit(X_train_tfidf, y_train)


# 5. Смотрим, что получилось после обучения.
print("\nОбучение завершено.")
print("Выполнено эпох:", mlp.n_iter_)

# Первая матрица весов соединяет входные признаки
# с 32 скрытыми нейронами: (число признаков, 32).
print("Форма весов первого слоя:", mlp.coefs_[0].shape)

# Вторая матрица соединяет 32 скрытых нейрона
# с одним выходом для бинарной классификации: (32, 1).
print("Форма весов выходного слоя:", mlp.coefs_[1].shape)

# Это значение одного обученного веса, а не вероятность.
# Вес может быть отрицательным или больше единицы.
print("Пример одного веса:", round(mlp.coefs_[0][0, 0], 4))

# Accuracy на части обучающих данных, автоматически отложенной
# для ранней остановки. Это ещё не результат на тестовой выборке.
print(
    "Лучшая accuracy на внутренней проверке:",
    round(mlp.best_validation_score_, 4),
)
```

```python
# ШАГ 13 — Оценка модели на тестовой выборке

# Импортируем функции для расчёта качества и построения матрицы ошибок.
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
)

# Библиотека для отображения графика.
import matplotlib.pyplot as plt


# 1. Превращаем тестовые тексты в числовые признаки.
# Используем ТОЛЬКО transform: словарь и веса IDF уже были
# построены на обучающей выборке в шаге 10.
X_test_tfidf = vectorizer.transform(test_df["text_clean"])

# Настоящие метки тестовых сообщений хранятся отдельно.
y_test = test_df["label"]

print("Тестовых сообщений:", X_test_tfidf.shape[0])
print("Признаков у сообщения:", X_test_tfidf.shape[1])


# 2. Просим обученную MLP определить тональность тестовых сообщений.
# Модель возвращает массив предсказанных меток 0 или 1.
y_pred = mlp.predict(X_test_tfidf)


# 3. Считаем общую долю правильных ответов по обоим классам.
accuracy = accuracy_score(y_test, y_pred)
print("\nAccuracy по всем тестовым сообщениям:", round(accuracy, 4))


# 4. Считаем Precision, Recall и F1 отдельно для негатива и позитива.
# pos_label указывает, какой класс сейчас считаем интересующим нас.
metric_rows = []

for class_id, class_name in [(0, "Негатив"), (1, "Позитив")]:
    metric_rows.append({
        "Класс": f"{class_name} ({class_id})",
        "Сообщений в тесте": int((y_test == class_id).sum()),
        "Precision": precision_score(
            y_test, y_pred, pos_label=class_id, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, pos_label=class_id, zero_division=0
        ),
        "F1": f1_score(
            y_test, y_pred, pos_label=class_id, zero_division=0
        ),
    })

metrics_table = pd.DataFrame(metric_rows).set_index("Класс")

print("\nМетрики для каждого класса:")
display(metrics_table.round(4))


# 5. Строим матрицу ошибок.
# Строки — настоящие метки; столбцы — предсказания модели.
# Порядок классов явно задаём как 0, затем 1.
cm = confusion_matrix(y_test, y_pred, labels=[0, 1])

print("Всего сообщений в матрице ошибок:", cm.sum())
print("Негатив ошибочно назван позитивом (FP):", cm[0, 1])
print("Позитив ошибочно назван негативом (FN):", cm[1, 0])

display_plot = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Негатив (0)", "Позитив (1)"],
)
display_plot.plot(cmap="Blues", values_format="d")
plt.title("Матрица ошибок на тестовой выборке")
plt.tight_layout()
plt.show()
```

```python
# ШАГ 14 — Просмотр ошибок модели на тестовых сообщениях

# Создаём таблицу для анализа.
# copy() сохраняет исходную test_df без изменений.
test_results = test_df[["ttext", "text_clean", "label"]].copy()

# Добавляем ответ модели из шага 13.
# Порядок y_pred совпадает с порядком строк test_df.
test_results["predicted_label"] = y_pred

# Узнаём, какому столбцу predict_proba соответствует класс 1.
# Это надёжнее, чем предполагать номер столбца.
positive_column = list(mlp.classes_).index(1)

# predict_proba возвращает вероятности классов для каждого сообщения.
# Берём вероятность позитива (класса 1).
test_results["p_positive"] = mlp.predict_proba(
    X_test_tfidf
)[:, positive_column]

# Группа FP: настоящий негатив (0) ошибочно назван позитивом (1).
false_positive = test_results.loc[
    (test_results["label"] == 0)
    & (test_results["predicted_label"] == 1)
]

# Группа FN: настоящий позитив (1) ошибочно назван негативом (0).
false_negative = test_results.loc[
    (test_results["label"] == 1)
    & (test_results["predicted_label"] == 0)
]

print("Негатив ошибочно назван позитивом (FP):", len(false_positive))
print("Позитив ошибочно назван негативом (FN):", len(false_negative))

# Столбцы, которые хотим видеть в примерах.
columns_to_show = [
    "ttext",
    "text_clean",
    "label",
    "predicted_label",
    "p_positive",
]

# Временно разрешаем Jupyter показывать длинные тексты целиком.
with pd.option_context("display.max_colwidth", None):
    print("\nПримеры FP — модель ошибочно увидела позитив:")
    display(
        false_positive[columns_to_show]
        .sample(n=min(5, len(false_positive)), random_state=42)
        .round({"p_positive": 3})
    )

    print("\nПримеры FN — модель ошибочно увидела негатив:")
    display(
        false_negative[columns_to_show]
        .sample(n=min(5, len(false_negative)), random_state=42)
        .round({"p_positive": 3})
    )
```

```python
# ШАГ 15 — Подробный разбор ошибок модели с 32 нейронами

# На шаге 14 мы создали test_results: исходный текст, очищенный текст,
# правильную метку, ответ модели и вероятность позитива.
# Берём только те строки, где ответ модели не совпал с правильной меткой.
errors_32 = test_results.loc[
    test_results["label"] != test_results["predicted_label"]
].copy()

# Отмечаем скобочные смайлики в ИСХОДНОМ тексте.
# Эти отметки нужны для анализа; теперь модель также видит скобки в text_clean.
errors_32["есть_)))"] = errors_32["ttext"].str.contains(
    r"\){2,}", regex=True, na=False
)
errors_32["есть_((("] = errors_32["ttext"].str.contains(
    r"\({2,}", regex=True, na=False
)

# FP: настоящий негатив (0) модель назвала позитивом (1).
false_positive_32 = errors_32.loc[
    (errors_32["label"] == 0)
    & (errors_32["predicted_label"] == 1)
]

# FN: настоящий позитив (1) модель назвала негативом (0).
false_negative_32 = errors_32.loc[
    (errors_32["label"] == 1)
    & (errors_32["predicted_label"] == 0)
]

print("Всего ошибок:", len(errors_32))
print("FP — негатив ошибочно назван позитивом:", len(false_positive_32))
print("FN — позитив ошибочно назван негативом:", len(false_negative_32))

# Показываем одни и те же столбцы во всех четырёх таблицах.
columns_to_analyze = [
    "ttext",             # Исходное сообщение: здесь видны скобки и смайлики.
    "text_clean",        # Очищенный текст: именно его получила модель.
    "label",             # Правильная метка из CSV.
    "predicted_label",   # Ответ модели с 32 нейронами.
    "p_positive",        # Вероятность позитива по оценке модели.
    "есть_)))",          # Есть ли позитивные скобки в исходном тексте.
    "есть_(((",          # Есть ли негативные скобки в исходном тексте.
]

# В FP модель ответила «позитив».
# Чем выше p_positive, тем увереннее был этот ошибочный ответ.
fp_confident = false_positive_32.sort_values(
    "p_positive", ascending=False
).head(20)

# В FP вероятность чуть выше 0,5 означает близость к порогу.
fp_near_threshold = false_positive_32.sort_values(
    "p_positive", ascending=True
).head(20)

# В FN модель ответила «негатив».
# Чем НИЖЕ p_positive, тем увереннее был ошибочный ответ «негатив».
fn_confident = false_negative_32.sort_values(
    "p_positive", ascending=True
).head(20)

# В FN вероятность чуть ниже 0,5 означает близость к порогу.
fn_near_threshold = false_negative_32.sort_values(
    "p_positive", ascending=False
).head(20)

# Временно показываем длинные сообщения полностью.
# Четыре таблицы содержат по 20 сообщений каждая.
with pd.option_context("display.max_colwidth", None):
    print("\n20 самых уверенных FP — негатив назван позитивом:")
    display(fp_confident[columns_to_analyze].round({"p_positive": 4}))

    print("\n20 FP, ближайших к порогу 0,5:")
    display(fp_near_threshold[columns_to_analyze].round({"p_positive": 4}))

    print("\n20 самых уверенных FN — позитив назван негативом:")
    display(fn_confident[columns_to_analyze].round({"p_positive": 4}))

    print("\n20 FN, ближайших к порогу 0,5:")
    display(fn_near_threshold[columns_to_analyze].round({"p_positive": 4}))
```

```python
# ШАГ 16 — Лемматизация: приводим слова к начальной форме

# pymorphy3 — морфологический анализатор русского языка.
# Например: «хорошая» и «хорошего» могут стать «хороший».
import pymorphy3
from functools import lru_cache

# Создаём анализатор один раз, а не заново для каждого слова.
morph = pymorphy3.MorphAnalyzer(lang="ru")

# Многие слова повторяются в разных твитах.
# Кеш хранит результат для уже встречавшихся слов и ускоряет обработку.
@lru_cache(maxsize=100_000)
def word_to_lemma(word):
    # parse() возвращает варианты разбора слова.
    # Берём наиболее вероятный вариант и его начальную форму.
    return morph.parse(word)[0].normal_form


def lemmatize_text(text):
    # Обрабатываем только последовательности русских букв.
    # Цифры, пробелы и смайлы ))) / ((( остаются без изменений.
    return re.sub(
        r"[а-яё]+",
        lambda match: word_to_lemma(match.group(0)),
        text,
    )

# Проверяем понятный пример до обработки всех сообщений.
example = "хорошая погода и хорошие друзья )))"
print("До лемматизации:", example)
print("После лемматизации:", lemmatize_text(example))
assert lemmatize_text(example).endswith(")))")

# Создаём НОВЫЙ столбец, сохраняя прежний text_clean.
# Разделение на train_df и test_df уже сделано на шаге 09:
# состав обеих выборок теперь не меняется.
train_df["text_lemma"] = train_df["text_clean"].map(lemmatize_text)
test_df["text_lemma"] = test_df["text_clean"].map(lemmatize_text)

print("Обучающих сообщений:", len(train_df))
print("Тестовых сообщений:", len(test_df))
print("Слов, взятых из кеша:", word_to_lemma.cache_info().hits)

# Показываем несколько текстов до и после лемматизации.
display(train_df[["text_clean", "text_lemma", "label"]].head(5))
```

```python
# ШАГ 17 — TF-IDF и MLP для лемматизированных текстов

# clone создаёт НОВЫЙ векторизатор с теми же настройками,
# но без словаря и весов IDF старого векторизатора.
# Поэтому обучение словаря вновь идёт только по train_df.
from sklearn.base import clone
lemma_vectorizer = clone(vectorizer)

# Обучаем новый словарь на лемматизированных обучающих текстах.
X_train_lemma_tfidf = lemma_vectorizer.fit_transform(
    train_df["text_lemma"]
)

# Тестовые тексты не изменяют словарь: для них только transform.
X_test_lemma_tfidf = lemma_vectorizer.transform(
    test_df["text_lemma"]
)

print("Обучающих сообщений:", X_train_lemma_tfidf.shape[0])
print("Тестовых сообщений:", X_test_lemma_tfidf.shape[0])
print("Признаков после лемматизации:", X_train_lemma_tfidf.shape[1])

# clone создаёт НОВУЮ MLP с прежними параметрами (32 нейрона,
# ReLU, Adam, early_stopping и random_state=42), но без обученных весов.
# Исходные vectorizer и mlp остаются для честного сравнения.
lemma_mlp = clone(mlp)
lemma_mlp.fit(X_train_lemma_tfidf, y_train)

print("Обучение модели с лемматизацией завершено.")
print("Выполнено эпох:", lemma_mlp.n_iter_)
```

```python
# ШАГ 18 — Сравнение моделей на ОДНИХ И ТЕХ ЖЕ тестовых сообщениях

# Старый ответ модели без лемматизации сохранён на шаге 13 в y_pred.
# Новая модель получает только лемматизированные признаки теста.
lemma_pred = lemma_mlp.predict(X_test_lemma_tfidf)

# Проверяем, что ни одна тестовая строка не была добавлена или потеряна.
assert len(y_test) == len(y_pred) == len(lemma_pred) == len(test_df)

# Считаем метрики отдельно для обеих моделей и обоих классов.
comparison_rows = []
for model_name, predictions in [
    ("Без лемматизации", y_pred),
    ("С лемматизацией", lemma_pred),
]:
    for class_id, class_name in [(0, "Негатив"), (1, "Позитив")]:
        comparison_rows.append({
            "Модель": model_name,
            "Класс": f"{class_name} ({class_id})",
            "Accuracy всех сообщений": accuracy_score(y_test, predictions),
            "Precision": precision_score(
                y_test, predictions, pos_label=class_id, zero_division=0
            ),
            "Recall": recall_score(
                y_test, predictions, pos_label=class_id, zero_division=0
            ),
            "F1": f1_score(
                y_test, predictions, pos_label=class_id, zero_division=0
            ),
        })

comparison_table = pd.DataFrame(comparison_rows)
print("Сравнение на", len(y_test), "одних и тех же тестовых сообщениях:")
display(comparison_table.round(4))

baseline_accuracy = accuracy_score(y_test, y_pred)
lemma_accuracy = accuracy_score(y_test, lemma_pred)
print("Изменение Accuracy, процентных пунктов:",
      round((lemma_accuracy - baseline_accuracy) * 100, 2))

# Считаем случаи, в которых новый подход исправил старую ошибку,
# и случаи, в которых он испортил прежний правильный ответ.
baseline_correct = (y_pred == y_test.to_numpy())
lemma_correct = (lemma_pred == y_test.to_numpy())
print("Исправлено прежних ошибок:", int((~baseline_correct & lemma_correct).sum()))
print("Новых ошибок:", int((baseline_correct & ~lemma_correct).sum()))

# Матрица ошибок для новой модели.
lemma_cm = confusion_matrix(y_test, lemma_pred, labels=[0, 1])
print("FP новой модели:", int(lemma_cm[0, 1]))
print("FN новой модели:", int(lemma_cm[1, 0]))
ConfusionMatrixDisplay(
    confusion_matrix=lemma_cm,
    display_labels=["Негатив (0)", "Позитив (1)"],
).plot(cmap="Blues", values_format="d")
plt.title("Матрица ошибок: модель с лемматизацией")
plt.tight_layout()
plt.show()
```

## ШАГ 19 — Какие слова влияют на ответ модели?

Анализируем основную модель **без лемматизации** (`mlp`, `X_train_tfidf`, `feature_names`), показавшую Accuracy **0,8036**. Слово связано со всеми 32 скрытыми нейронами, а каждый нейрон — с выходом модели. Поэтому отдельный вес первого слоя ещё нельзя назвать «весом позитива» или «весом негатива».

Для каждого обучающего сообщения смотрим, какие нейроны включились после ReLU. Затем для каждого нейрона считаем долю таких сообщений. Влияние признака оцениваем как сумму произведений: **вес признака → нейрон × вес нейрона → выход × доля включений нейрона**. Положительный результат означает средний сдвиг выходного сигнала к классу `1`, отрицательный — к классу `0`.

Это *средняя чувствительность выходного сигнала к TF-IDF-признаку*, а не вероятность и не универсальный «смысл» слова. В конкретном твите влияние зависит от того, какие нейроны включились и каков TF-IDF-вес слова. Рейтинг рассчитывается только по обучающим сообщениям; тестовые тексты не участвуют.


```python
# ШАГ 19 — Расчёт средних направленных влияний слов

# Анализируем именно модель БЕЗ лемматизации.
# mlp была обучена на X_train_tfidf в шаге 12.
# feature_names получены из её TF-IDF-словаря в шаге 10.
import numpy as np

# В бинарной модели класс 1 — позитив, класс 0 — негатив.
assert list(mlp.classes_) == [0, 1]

# Матрица первого слоя: строки — признаки, столбцы — 32 нейрона.
# Например, first_layer[j, k] — связь слова j с нейроном k.
first_layer = mlp.coefs_[0]

# Вектор выходного слоя: связь каждого из 32 нейронов
# с единственным выходным сигналом для класса 1.
output_weights = mlp.coefs_[1][:, 0]

# Проверяем размеры, чтобы не сопоставить веса с чужим словарём.
assert first_layer.shape == (len(feature_names), len(output_weights))

# Считаем входы скрытых нейронов для ОБУЧАЮЩИХ сообщений.
# Умножение разреженной матрицы TF-IDF на первый слой даёт
# обычную матрицу: сообщения × 32 нейрона.
# intercepts_[0] — добавочные свободные члены 32 нейронов.
hidden_before_relu = X_train_tfidf @ first_layer + mlp.intercepts_[0]

# ReLU оставляет положительное значение и превращает остальные в 0.
# Для каждого нейрона считаем долю сообщений, где он включился.
# Например, 0.70 означает: нейрон активен на 70% обучающих текстов.
active_fraction = np.asarray(hidden_before_relu > 0).mean(axis=0)

# Один признак имеет 32 связи с нейронами.
# Умножаем каждую связь на вес нейрона к выходу и на долю включений.
# Сумма этих 32 произведений даёт одно среднее направленное влияние.
# Форма influence: (число признаков,).
influence = first_layer @ (output_weights * active_fraction)

# Считаем, в скольких обучающих сообщениях встретился признак.
# Это поможет заметить слова, попавшие в рейтинг из-за редкости.
document_count = np.asarray((X_train_tfidf > 0).sum(axis=0)).ravel()

# Сопоставляем названия признаков TF-IDF с вычисленными числами.
influence_table = pd.DataFrame({
    "признак": feature_names,
    "среднее_влияние": influence,
    "сообщений_с_признаком": document_count,
})

# В задании нужны именно слова. Поэтому числа и скобки
# не включаем в топ-20 слов, но смайлы покажем отдельно.
word_influence = influence_table.loc[
    influence_table["признак"].str.fullmatch(r"[а-яё]+")
]

# Наибольшие положительные значения сдвигают ответ к позитиву.
top_positive_words = word_influence.nlargest(20, "среднее_влияние")

# Наиболее отрицательные значения сдвигают ответ к негативу.
top_negative_words = word_influence.nsmallest(20, "среднее_влияние")

print("Топ-20 слов: сдвиг в сторону позитива (класс 1)")
display(top_positive_words.round({"среднее_влияние": 4}))

print("Топ-20 слов: сдвиг в сторону негатива (класс 0)")
display(top_negative_words.round({"среднее_влияние": 4}))

# Смайлы — отдельные признаки, которые часто помогают с тональностью.
smile_influence = influence_table.loc[
    influence_table["признак"].str.fullmatch(r"\)+|\(+")
]
print("Отдельно: влияние скобочных смайлов")
display(smile_influence.sort_values("среднее_влияние").round({"среднее_влияние": 4}))
```

## Вывод по наиболее влиятельным признакам

Веса первого слоя имеют форму **(39 709, 32)**: у каждого TF-IDF-признака есть связь с каждым из 32 скрытых нейронов. Веса выходного слоя показывают, как эти нейроны воздействуют на ответ. Для слова *j* мы вычислили среднее направленное влияние как сумму по нейронам: `вес слова→нейрон × вес нейрона→выход × доля сообщений, где нейрон активен`. Доля активности рассчитана по **обучающей** выборке, без просмотра тестовых текстов. Плюс означает сдвиг к позитиву, минус — к негативу. Это чувствительность выходного сигнала, **не вероятность класса**.

Списки в основном логичны: в сторону позитива ведут `приятно`, `обожаю`, `спасибо`, `ура`, `счастлива`; в сторону негатива — `обидно`, `жаль`, `грустно`, `печально`, `ненавижу`. В топе позитива несколько вариантов смеха (`ахах`, `ахаха`, `ахахах`), поскольку TF-IDF считает их разными словами. Встречаются контекстные или случайные связи: `снегу` и `главное` не всегда позитивны, а `шерлока` само по себе не негативно. Столбец «сообщений с признаком» помогает видеть, насколько часто слово встречалось в обучении. Число слева от таблицы — индекс признака в исходном словаре, а не место слова в рейтинге.

Скобочные смайлы дают особенно сильные сигналы: `))` имеет среднее влияние **+13,29**, а `((` — **−13,63**. Модель использует эти регулярные связи в твитах, но по ним нельзя судить о понимании смысла текста. В конкретной фразе итог зависит и от других слов, и от того, какие нейроны включит ReLU; поэтому таблицы не задают универсальную тональность каждого слова.

Вероятности выводим для **обоих классов** из `predict_proba`. Это оценки модели, а не гарантия правильности ответа. Последний столбец показывает лишь совпадение с оценкой автора.


```python
# ШАГ 20 — Семь собственных фраз и вероятности обоих классов

# Каждая строка содержит исходную фразу и ВАШУ оценку.
# Эти метки НЕ передаются модели при предсказании.
manual_examples = [
    {"phrase": "Ну вы мужики блин даете!!!  )))", "human_label": 1},
    {"phrase": "На улице стоял осенний хмурый день", "human_label": 0},
    {"phrase": "Не парьте мне мозг коллеги.", "human_label": 0},
    {"phrase": "Ничего не может быть лучше чем яркий солнечный день в отпуске", "human_label": 1},
    {"phrase": "Казнить нельзя помиловать", "human_label": 0},
    {"phrase": "О сколько нам открытий чудных готовит просвященья дух и опыт сын ошибок трудных и гений парадоксов друг", "human_label": 1},
    {"phrase": "Тру-ля-ля, аха-ха, никчемная ложечка ))))", "human_label": 1},
]

# Создаём таблицу: одна строка — одна ваша фраза.
manual_results = pd.DataFrame(manual_examples)
manual_results.index = range(1, len(manual_results) + 1)
manual_results.index.name = "№ фразы"

# Используем ТУ ЖЕ функцию очистки, что и при обучении.
# Смайлы ))) и (((( сохраняются по правилам шага 06.
manual_results["text_clean"] = manual_results["phrase"].map(clean_text)

# Векторизатор УЖЕ обучен на train_df в шаге 10.
# transform применяет готовый словарь и IDF к новым фразам;
# fit или fit_transform здесь вызвали бы утечку и несовместимость.
X_manual_tfidf = vectorizer.transform(manual_results["text_clean"])

# Проверяем, что число столбцов совпало с входом MLP.
assert X_manual_tfidf.shape == (len(manual_results), X_train_tfidf.shape[1])

# Получаем ответ основной MLP: 0 — негатив, 1 — позитив.
manual_results["predicted_label"] = mlp.predict(X_manual_tfidf)

# predict_proba возвращает ДВА столбца вероятностей.
# Номера столбцов определяем по mlp.classes_, а не угадываем.
class_probabilities = mlp.predict_proba(X_manual_tfidf)
negative_column = list(mlp.classes_).index(0)
positive_column = list(mlp.classes_).index(1)
manual_results["p_negative"] = class_probabilities[:, negative_column]
manual_results["p_positive"] = class_probabilities[:, positive_column]

# Сравниваем ответ модели с вашей субъективной оценкой.
# Эта проверка не меняет обученную модель и её тестовую Accuracy.
manual_results["совпало_с_вашей_оценкой"] = (
    manual_results["predicted_label"] == manual_results["human_label"]
)

print("Ручная проверка: 0 — негатив, 1 — позитив")
print("Совпадений с вашей оценкой:",
      int(manual_results["совпало_с_вашей_оценкой"].sum()),
      "из", len(manual_results))

# Показываем длинные тексты полностью и округляем только отображение.
with pd.option_context("display.max_colwidth", None):
    display(manual_results[[
        "phrase", "text_clean", "human_label", "predicted_label",
        "p_negative", "p_positive", "совпало_с_вашей_оценкой",
    ]].round({"p_negative": 4, "p_positive": 4}))
```

**Граница подхода.** TF-IDF превращает текст в числа по частотам отдельных токенов, а MLP подбирает веса для уменьшения ошибки на размеченных примерах. Для улучшения этой работы полезно отдельно проверять качество разметки, неоднозначные твиты и примеры, где порядок слов меняет смысл.
